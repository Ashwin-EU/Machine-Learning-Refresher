# Set up the validation framework

```python
n_val = int(len(df) * 0.2)
n_test = int(len(df) * 0.2)
n_train = len(df) - n_val - n_test
```

```python
idx = np.arange(len(df))
np.random.shuffle(idx)
df_val = df.iloc[idx[:n_val]]
df_test = df.iloc[idx[n_val:n_val+n_test]]
df_train = df.iloc[idx[n_val+n_test:]]
```

```python
df_train.head()
```

```python
df_train = df_train.reset_index(drop=True)
df_val = df_val.reset_index(drop=True)
df_test = df_test.reset_index(drop=True)

y_train = np.log1p(df_train.msrp.values)
y_val = np.log1p(df_val.msrp.values)
y_test = np.log1p(df_test.msrp.values)

del df_train["msrp"]
del df_val["msrp"]
del df_test["msrp"]
```

# Train Linear Regression

```python
def train_linear_regression(X, y):
    X = np.array(X)
    y = np.array(y)
    ones = np.ones(X.shape[0])
    X = np.column_stack([ones, X])
    XTX = X.T.dot(X)
    XTX_inv = np.linalg.inv(XTX)
    w_full = XTX_inv.dot(X.T).dot(y)
    return w_full[0], w_full[1:]
```

```python
df_train.dtypes
```

# Feature Engineering

```python
base = ['engine_hp', 'engine_cylinders', 'highway_mpg', 'city_mpg', 'popularity']
```

```python
def prepare_X(df):
    df = df.copy()
    df['age'] = 2026 - df.year
    features = base + ['age']
    df_num = df[features].fillna(0)
    X = df_num.values
    return X
X_train = prepare_X(df_train)
```

```python
w0, w = train_linear_regression(X_train, y_train)
```

```python
y_pred_train = w0 + X_train.dot(w)
```

```python
X_val = prepare_X(df_val)
y_pred_val = w0 + X_val.dot(w)
```

```python
sns.histplot(y_pred_train, bins=50, color='red', label='predicted')
sns.histplot(y_train, bins=50, color='blue', label='actual')
plt.legend()
```

```python
def rmse(y, y_pred):
    error = y - y_pred
    mse = (error ** 2).mean()
    return np.sqrt(mse)
```

```python
rmse(y_train, y_pred_train)
```

```python
rmse(y_val, y_pred_val)
```

# Categorical Variables

```python
from sklearn.preprocessing import OneHotEncoder
```

```python
df.dtypes
```

```python
from sklearn.impute import SimpleImputer
from sklearn.preprocessing import OneHotEncoder
from sklearn.compose import ColumnTransformer
def add_features(df):
    df = df.copy()   
    df['age'] = 2026 - df.year
    return df
def prepare_X(df_train):
    df_train = add_features(df_train)
    categorical_columns = df_train.dtypes[df_train.dtypes == "str"].index.tolist()
    numerical_columns = df_train.dtypes[df_train.dtypes != "str"].index.tolist() 
    preprocessor = ColumnTransformer(
        transformers=[
            ('num', SimpleImputer(strategy='constant'), numerical_columns),
            ('cat', OneHotEncoder(handle_unknown='ignore'), categorical_columns)
        ]
    )
    X_train = preprocessor.fit_transform(df_train)
    return X_train, preprocessor
X_train, preprocessor = prepare_X(df_train)
X_val = add_features(df_val)
X_val = preprocessor.transform(X_val)
```

```python
X_train.shape
```

```python
X_val.shape
```

# Regularized Regression

```python
from sklearn.linear_model import LinearRegression, Ridge, Lasso, ElasticNet
model = LinearRegression()
model.fit(X_train, y_train)
y_pred_train = model.predict(X_train)
y_pred_val = model.predict(X_val)
```

```python
rmse(y_train, y_pred_train)
```

```python
rmse(y_val, y_pred_val)
```

```python
model = Ridge(alpha=0.1)
model.fit(X_train, y_train)
y_pred_train = model.predict(X_train)
y_pred_val = model.predict(X_val)
```

```python
rmse(y_train, y_pred_train)
```

```python
rmse(y_val, y_pred_val)
```

```python
model = Lasso(alpha=0.1)
model.fit(X_train, y_train)
y_pred_train = model.predict(X_train) 
y_pred_val = model.predict(X_val)
```

```python
rmse(y_train, y_pred_train)
```

```python
rmse(y_val, y_pred_val)
```

```python
model = ElasticNet(alpha=0.3, l1_ratio=0.5)
model.fit(X_train, y_train)
y_pred_train = model.predict(X_train)   
y_pred_val = model.predict(X_val)
```

```python
rmse(y_train, y_pred_train)
```

```python
rmse(y_val, y_pred_val)
```

# Tuning the Model

```python
alphas = np.arange(0.01, 1, 0.01)
alphas
```

```python
results = []
for alpha in alphas:
    model = Ridge(alpha=alpha)
    model.fit(X_train, y_train)
    train_rmse = rmse(y_train, model.predict(X_train))
    val_rmse = rmse(y_val, model.predict(X_val))
    results.append({
        "alpha": alpha,
        "train_rmse": train_rmse,
        "val_rmse": val_rmse        
    })
    results_df = pd.DataFrame(results)
results_df = results_df.sort_values(by="val_rmse", ascending=True).reset_index(drop=True)
results_df
```

# Test set or Holdout set validation with best alpha

```python
df_full_train = pd.concat([df_train, df_val]).reset_index(drop=True)
y_full_train = np.concatenate([y_train, y_val])
model = Ridge(alpha=0.05)
X_full_train, preprocessor = prepare_X(df_full_train)
model.fit(X_full_train, y_full_train)
rmse(y_full_train, model.predict(X_full_train))
```

```python
X_full_test = add_features(df_test)
X_full_test = preprocessor.transform(X_full_test)
y_full_test = y_test
rmse(y_full_test, model.predict(X_full_test))
```
